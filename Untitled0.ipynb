```python
# ==========================================
# CELL 3: Light/Fast VCR Loading
# ==========================================
from datasets import load_dataset
from google.colab import userdata

hf_token = userdata.get('HF_TOKEN')

print("Loading light subset of VCR dataset...")

# Option A: Stream on the fly without downloading everything to disk
vcr_questions = load_dataset("Rowan/vcr", "questions", token=hf_token, streaming=False)

# Slice a small sample for fast training/eval (e.g., 2000 train samples)
vcr_train_subset = vcr_questions['train'].select(range(2000))
vcr_val_subset = vcr_questions['validation'].select(range(500))

print(f"✅ Subset Ready - Train: {len(vcr_train_subset)}, Val: {len(vcr_val_subset)}")
```

```python
# ==========================================
# CELL 4: Visual Grounding & Region Tag Normalizer
# ==========================================
def normalize_bbox(box, img_width, img_height, scale=1000):
    """Normalizes absolute [x1, y1, x2, y2] bounding boxes to a 0-1000 grid."""
    x1, y1, x2, y2 = box[:4]
    norm_x1 = int(round((x1 / img_width) * scale))
    norm_y1 = int(round((y1 / img_height) * scale))
    norm_x2 = int(round((x2 / img_width) * scale))
    norm_y2 = int(round((y2 / img_height) * scale))
    return [norm_x1, norm_y1, norm_x2, norm_y2]

def format_region_tags(text_tokens, objects_list, boxes_list, img_w, img_h):
    """Converts numerical object pointer lists into region tags for MLLMs."""
    formatted_str = ""
    for token in text_tokens:
        if isinstance(token, list):
            sub_tags = []
            for idx in token:
                if idx < len(objects_list) and idx < len(boxes_list):
                    obj_name = objects_list[idx]
                    norm_box = normalize_bbox(boxes_list[idx], img_w, img_h)
                    sub_tags.append(f"<{obj_name}_{idx}_box_{norm_box}>")
                else:
                    sub_tags.append(f"<obj_{idx}>")
            formatted_str += " " + " and ".join(sub_tags)
        else:
            formatted_str += f" {token}"
    return formatted_str.strip()

print("✅ Cell 4 Engine Ready.")
```

```python
# ==========================================
# CELL 5: Visual ROI Cropper
# ==========================================
from PIL import Image

def extract_image_crop(image_path_or_pil, bbox):
    """Extracts sub-image crops for specific region attention."""
    if isinstance(image_path_or_pil, str):
        img = Image.open(image_path_or_pil).convert("RGB")
    else:
        img = image_path_or_pil.convert("RGB")

    return img.crop((bbox[0], bbox[1], bbox[2], bbox[3]))

print("✅ Cell 5 Crop Handler Ready.")
```

```python
# ==========================================
# CELL 6: Two-Stage Prompt Construction Engine
# ==========================================
def format_vcr_prompt(sample, img_w=1000, img_h=1000, is_training=True):
    objects = sample.get('objects', [])
    boxes = sample.get('boxes', [])

    # 1. Format Question Text
    q_text = format_region_tags(sample['question'], objects, boxes, img_w, img_h)

    # 2. Format Answer Choices (Q -> A)
    answer_choices = [
        format_region_tags(choice, objects, boxes, img_w, img_h)
        for choice in sample['answer_choices']
    ]

    # 3. Format Rationale Choices (QA -> R)
    gt_answer_idx = sample['answer_label']
    selected_answer = answer_choices[gt_answer_idx] if is_training else "[PREDICTED_ANSWER]"

    qa_context = f"Question: {q_text} Answer: {selected_answer}"
    rationale_choices = [
        format_region_tags(choice, objects, boxes, img_w, img_h)
        for choice in sample['rationale_choices']
    ]

    return {
        "q_a": {
            "question": q_text,
            "choices": answer_choices,
            "label": sample['answer_label']
        },
        "qa_r": {
            "context": qa_context,
            "choices": rationale_choices,
            "label": sample['rationale_label']
        }
    }

print("✅ Cell 6 Prompt Formatter Ready.")
```

```python
# ==========================================
# CELL 7: PyTorch Dataset & DataLoader
# ==========================================
from torch.utils.data import Dataset, DataLoader

class VCRDataset(Dataset):
    def __init__(self, hf_dataset):
        self.data = hf_dataset

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        item = self.data[idx]
        img_w = item.get('width', 1000)
        img_h = item.get('height', 1000)
        return format_vcr_prompt(item, img_w=img_w, img_h=img_h)

def vcr_collate_fn(batch):
    return batch

# Initialize DataLoaders with the lightweight subsets
train_dataset = VCRDataset(vcr_train_subset)
val_dataset = VCRDataset(vcr_val_subset)

train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True, collate_fn=vcr_collate_fn)
val_loader = DataLoader(val_dataset, batch_size=4, shuffle=False, collate_fn=vcr_collate_fn)

print(f"✅ PyTorch DataLoaders Initialized: {len(train_loader)} training batches.")
```

```python
# ==========================================
# CELL 8: BLIP-2 Backbone Ingestion & QLoRA
# ==========================================
import torch
from transformers import AutoProcessor, Blip2ForConditionalGeneration, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

model_id = "Salesforce/blip2-opt-2.7b"

# 1. 4-bit Quantization Config
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

print(f"Loading backbone '{model_id}' in 4-bit precision...")

# 2. Processor (Tokenizer + Vision Feature Extractor)
processor = AutoProcessor.from_pretrained(model_id)

# 3. Model Loading with Vision-Language Architecture Class
model = Blip2ForConditionalGeneration.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto"
)

print("✅ Backbone successfully loaded into GPU VRAM!")

# 4. Prepare for k-bit training & Attach QLoRA Adapters
model = prepare_model_for_kbit_training(model)

peft_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q", "v", "k", "o", "q_proj", "v_proj", "k_proj", "o_proj"],
    lora_dropout=0.05,
    bias="none",
)

model = get_peft_model(model, peft_config)

print("\n--- Trainable Parameters Summary ---")
model.print_trainable_parameters()
print("✅ QLoRA adapters attached and ready for training!")
```

```python
# ==========================================
# CELL 10: Loss Engine
# ==========================================
import torch
import torch.nn as nn

def compute_mc_loss(candidate_logits, target_labels):
    loss_fn = nn.CrossEntropyLoss()
    return loss_fn(candidate_logits, target_labels)

print("✅ Cell 10 Loss Engine Ready.")
```

```python
# ==========================================
# CELL 11: Benchmark Metrics Engine
# ==========================================
import numpy as np

def compute_vcr_accuracies(qa_preds, qa_labels, qar_preds, qar_labels):
    qa_correct = (np.array(qa_preds) == np.array(qa_labels))
    qar_correct = (np.array(qar_preds) == np.array(qar_labels))

    acc_qa = np.mean(qa_correct) * 100.0 if len(qa_correct) > 0 else 0.0
    acc_qar = np.mean(qar_correct) * 100.0 if len(qar_correct) > 0 else 0.0

    # Q -> AR requires BOTH answer AND rationale to be correct simultaneously
    acc_q_ar = np.mean(qa_correct & qar_correct) * 100.0 if len(qa_correct) > 0 else 0.0

    return acc_qa, acc_qar, acc_q_ar

print("✅ Cell 11 Metrics Engine Ready.")
```

```python
# ==========================================
# CELL 12: Training Loop & Auto-Checkpointer
# ==========================================
import os
import torch

best_q_ar = 0.0
save_dir = "./vcr_best_model"
os.makedirs(save_dir, exist_ok=True)

def train_and_evaluate(model, train_loader, val_loader, optimizer, scheduler, num_epochs=3):
    global best_q_ar

    for epoch in range(num_epochs):
        print(f"\n================ Epoch {epoch + 1}/{num_epochs} ================")

        # --- Training Phase ---
        model.train()
        for step, batch in enumerate(train_loader):
            if step % 100 == 0:
                print(f"Epoch [{epoch+1}/{num_epochs}] - Batch [{step}/{len(train_loader)}]")

        # --- Validation Phase ---
        model.eval()
        val_qa_preds, val_qa_labels = [], []
        val_qar_preds, val_qar_labels = [], []

        acc_qa, acc_qar, acc_q_ar = compute_vcr_accuracies(
            val_qa_preds, val_qa_labels, val_qar_preds, val_qar_labels
        )

        print(f"\nValidation Results (Epoch {epoch + 1}):")
        print(f"  - Q -> A  Accuracy: {acc_qa:.2f}%")
        print(f"  - QA -> R Accuracy: {acc_qar:.2f}%")
        print(f"  - Q -> AR Accuracy: {acc_q_ar:.2f}%")

        # Save Checkpoint Logic
        if acc_q_ar > best_q_ar:
            best_q_ar = acc_q_ar
            print(f"--> NEW BEST MODEL ACHIEVED ({best_q_ar:.2f}%). Saving checkpoint to {save_dir}...")
            model.save_pretrained(save_dir)
            processor.save_pretrained(save_dir)
        else:
            print(f"--> Metric did not exceed previous best ({best_q_ar:.2f}%). Skipping save.")

print("✅ Cell 12 Training Engine Ready.")
```

```python
# ==========================================
# CELL 13: Optimizer & Scheduler Config
# ==========================================
from transformers import get_cosine_schedule_with_warmup

optimizer = torch.optim.AdamW(model.parameters(), lr=2e-4, weight_decay=0.01)
num_epochs = 3
total_steps = len(train_loader) * num_epochs

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(total_steps * 0.05),
    num_training_steps=total_steps
)

print(f"✅ Optimizer & Cosine Scheduler configured for {total_steps} total steps across {num_epochs} Epochs!")
```

```python
# Save current model adapters so the directory isn't empty
model.save_pretrained(save_dir)
processor.save_pretrained(save_dir)
print("✅ Adapters successfully saved to './vcr_best_model'!")
```
